structured text format (**CSV file**) and load it back — this is our **formatted text type data**.

## Question 2: Build a PIPELINE using Data Processing and Feature Engineering



### 2.1 Data Processing

```python
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# ---------- STEP 1 : DATA PROCESSING ----------
def simple_stem(word):
    """Light stemmer -> removes common suffixes (ing, ed, es, ly, s)."""
    for suf in ("ing", "ed", "es", "ly", "s"):
        if word.endswith(suf) and len(word) > len(suf) + 2:
            return word[: -len(suf)]
    return word

def preprocess(text):
    text = str(text).lower()
    text = re.sub(r"[^a-z\s]", " ", text)
    words = text.split()
    words = [w for w in words if w not in ENGLISH_STOP_WORDS]
    words = [simple_stem(w) for w in words]
    return " ".join(words)

df["clean_text"] = df["text"].apply(preprocess)

print("ORIGINAL  :", df["text"].iloc[0])
print("PROCESSED :", df["clean_text"].iloc[0])
print()
print("ORIGINAL  :", df["text"].iloc[21])
print("PROCESSED :", df["clean_text"].iloc[21])
```

### 2.2 Feature Engineering

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# ---------- STEP 2 : FEATURE ENGINEERING (text -> numbers) ----------
# (a) Bag of Words : count of each word in every document
bow_vectorizer = CountVectorizer()
bow_features = bow_vectorizer.fit_transform(df["clean_text"])

print("BAG OF WORDS")
print("Shape (documents x unique words):", bow_features.shape)
print("First 15 vocabulary words:", list(bow_vectorizer.get_feature_names_out()[:15]))
print("BoW vector of the 1st document:\n", bow_features[0].toarray())
print()

# (b) TF-IDF : gives higher weight to important/rare words (unigrams + bigrams)
tfidf_vectorizer = TfidfVectorizer(ngram_range=(1, 2))
tfidf_features = tfidf_vectorizer.fit_transform(df["clean_text"])

print("TF-IDF")
print("Shape (documents x features):", tfidf_features.shape)
print("Sample features:", list(tfidf_vectorizer.get_feature_names_out()[:8]))
print("TF-IDF vector of the 1st document (rounded):\n", np.round(tfidf_features[0].toarray(), 2))
```

### 2.3 Complete Pipeline (train, test, evaluate, predict)

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# ---------- STEP 3 : FULL PIPELINE (processing -> features -> model) ----------
# Split the formatted text data: 75% train, 25% test
X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"], test_size=0.25, random_state=15, stratify=df["label"]
)

# Pipeline : TF-IDF does processing + feature engineering, Naive Bayes classifies
pipe = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),   # feature engineering
    ("model", MultinomialNB()),                          # classification model
])

pipe.fit(X_train, y_train)          # TRAIN the pipeline
y_pred = pipe.predict(X_test)       # TEST the pipeline

print("Actual labels :", list(y_test))
print("Predictions   :", [str(p) for p in y_pred])
print()
print("Accuracy of the pipeline:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")
print()
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print()
print("Classification Report:\n", classification_report(y_test, y_pred))

# Use the pipeline to classify brand-new sentences
new_reviews = [
    "The movie was absolutely wonderful and I loved the acting",
    "This film was so boring and the story was terrible",
]
print("NEW REVIEWS :", new_reviews)
print("PREDICTIONS :", [str(p) for p in pipe.predict(new_reviews)])
```
